```python
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score, classification_report

iris = load_iris()
X_train, X_test, y_train, y_test = train_test_split(
    iris.data, iris.target, test_size=0.2, random_state=42, stratify=iris.target
)
print("Данные успешно загружены и разделены!")
```

```python
for k in (3, 5, 7):
    knn = make_pipeline(StandardScaler(), KNeighborsClassifier(n_neighbors=k))
    knn.fit(X_train, y_train)
    y_pred = knn.predict(X_test)

    print(f"=== KNN (n_neighbors={k}) ===")
    print("Accuracy:", round(accuracy_score(y_test, y_pred), 4))
    print(classification_report(y_test, y_pred, target_names=iris.target_names))
```

```python
trees = {
    "Дерево без ограничений": DecisionTreeClassifier(random_state=42),
    "Дерево с max_depth=2": DecisionTreeClassifier(max_depth=2, random_state=42)
}

for name, tree in trees.items():
    tree.fit(X_train, y_train)
    y_pred = tree.predict(X_test)

    print(f"=== {name} ===")
    print("Accuracy:", round(accuracy_score(y_test, y_pred), 4))
    print(classification_report(y_test, y_pred, target_names=iris.target_names))
```

```python
models_3 = {
    "Логистическая регрессия": make_pipeline(StandardScaler(), LogisticRegression(max_iter=1000)),
    "K ближайших соседей (k=5)": make_pipeline(StandardScaler(), KNeighborsClassifier(n_neighbors=5)),
    "Дерево решений": DecisionTreeClassifier(random_state=42)
}

table_data = []
for name, model in models_3.items():
    model.fit(X_train, y_train)
    acc = accuracy_score(y_test, model.predict(X_test))
    table_data.append({"Алгоритм": name, "Accuracy": round(acc, 4)})

df_results = pd.DataFrame(table_data).sort_values(by="Accuracy", ascending=False)
display(df_results) # Функция display красиво отрисует таблицу в Colab
```

```python
mlp = make_pipeline(StandardScaler(), MLPClassifier(max_iter=1000, random_state=42))
mlp.fit(X_train, y_train)
y_pred_mlp = mlp.predict(X_test)

print("=== Многослойный персептрон (MLPClassifier) ===")
print("Accuracy:", round(accuracy_score(y_test, y_pred_mlp), 4))
print(classification_report(y_test, y_pred_mlp, target_names=iris.target_names))
```
